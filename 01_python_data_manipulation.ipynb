```python
for num in range(2, 51):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

```python
text = "Artificial Intelligence"
vowels = "aeiouAEIOU"

count = 0
for char in text:
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```

```python
transactions = [1200, -250, 3000, -500, -1500, 800]

print("Withdrawals:")
for t in transactions:
    if t < 0:
        print(t)

print("Deposits:")
for t in transactions:
    if t > 0:
        print(t)
```

```python
employees = [("John", 35000), ("Sara", 55000), ("Mike", 72000), ("Lily", 29000)]

print("Employees earning more than 50000:")
for name, salary in employees:
    if salary > 50000:
        print(name)

print("Employees earning less than 40000:")
for name, salary in employees:
    if salary < 40000:
        print(name)
```

```python
transactions = [200, -500, 1500, -1200, 2500, -700, 100]

total_deposit = 0
total_withdrawal = 0

for t in transactions:
    if t > 0:
        total_deposit += t
    else:
        total_withdrawal += t

print("Total Deposits:", total_deposit)
print("Total Withdrawals:", total_withdrawal)
```

```python
num = 84
i = 2

while i <= num:
    if num % i == 0:
        print(i)
        num = num // i
    else:
        i += 1
```

```python
for i in range(1, 11):
    for j in range(1, 11):
        result = i * j
        if result % 2 == 0:
            print(f"{i} x {j} = {result}")
```
