```python
text = "mississippi"

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```

```python
country_capital = {"India": "Delhi", "France": "Paris", "Japan": "Tokyo"}

reverse = {}

for key in country_capital:
    value = country_capital[key]
    reverse[value] = key

print(reverse)
```

```python
stock1 = {"apple": 50, "banana": 30, "orange": 10}
stock2 = {"banana": 20, "orange": 15, "grape": 40}

result = stock1.copy()

for item in stock2:
    if item in result:
        result[item] += stock2[item]
    else:
        result[item] = stock2[item]

print(result)
```

```python
students = {
    "Alice": {"math": 90, "science": 85},
    "Bob": {"math": 70, "science": 75},
    "Charlie": {"math": 95, "science": 80}
}

# Highest math score
top = ""
max_marks = 0

for name in students:
    if students[name]["math"] > max_marks:
        max_marks = students[name]["math"]
        top = name

print("Top in Math:", top)

# Average science marks
total = 0

for name in students:
    total += students[name]["science"]

average = total / len(students)
print("Average Science:", average)

# Add English subject
for name in students:
    students[name]["english"] = 0

print(students)
```

```python
prices = {"pen": 10, "book": 50, "pencil": 5}
cart = ["pen", "book", "pen", "pencil"]

total = 0
bill = {}

for item in cart:
    total += prices[item]

    if item in bill:
        bill[item] += 1
    else:
        bill[item] = 1

print("Total =", total)
print("Detailed Bill =", bill)
```

```python
users = {"john":"12345", "alex":"Abc@123", "rita":"pass#2024"}

result = {}

for user in users:
    password = users[user]

    if len(password) < 6:
        strength = "Weak"
    elif password.isalnum():
        strength = "Medium"
    else:
        strength = "Strong"

    result[user] = strength

print(result)
```
